As shown, some features exhibit low Pearson correlation coefficient ($R_p$) but high importance (for example, `B_X+`). This demonstrates why the correlation coefficient is not always a reliable measure of feature importance. The Pearson correlation coefficient ($R_p$):

- Captures only linear relationships
- Is sensitive to outliers  

- Performs poorly with categorical or highly discrete features

- Ignores features interaction

For more robust feature importance assessment, model-based methods are recommended.

When features exhibit strong correlation, removing one from each correlated pair offers several benefits:

- it yields simpler and more robust models
- reduces data collection requirements

- and streamlines the analysis of feature importance.



````python
from scipy.stats import spearmanr



def get_correlated_features(X, threshold = 0.85):
    feature_names = X.columns

    # we use Spearman coefficient instead of Rp
    corr = spearmanr(X_train).correlation

    # Ensure the correlation matrix is symmetric
    corr = (corr + corr.T) / 2
    np.fill_diagonal(corr, 1)

    correlated_features = []
    for i in range(len(corr)):
        for j in range(i):
            if abs(corr[i, j]) > threshold:
                feature_name = feature_names[i]
                if feature_name not in correlated_features:
                    correlated_features.append(feature_name)
    return correlated_features


print(get_correlated_features(X_train, threshold=0.85))
````

````python
from sklearn.ensemble import RandomForestRegressor
from tqdm import tqdm
import matplotlib.pyplot as plt

X = data_new.drop(['PBE band gap'], axis = 1)
y = data_new['PBE band gap']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, shuffle=True, random_state=42)

correlated_features = get_correlated_features(X_train, threshold=0.85)
X_train_clean = X_train.drop(columns=correlated_features)

X_train_sub, X_val, y_train_sub, y_val = train_test_split(X_train_clean, y_train, test_size=0.2, shuffle=True, random_state=42)

features = list(X_train_clean.columns)
best_features = []
mse_history = []

num_features_to_select = 10

for k in tqdm(range(num_features_to_select), desc="Forward Selection"):
    best_score = np.inf
    current_best_feature = None

    for feature in features:
        if feature in best_features:
            continue

        current_features = best_features + [feature]
        X_step = X_train_sub[current_features].values

        model = RandomForestRegressor(n_estimators=10, n_jobs=-1, random_state=42)
        model.fit(X_step, y_train_sub)

        val_preds = model.predict(X_val[current_features].values)
        score = MSE(val_preds, y_val)

        if score < best_score:
            best_score = score
            current_best_feature = feature

    if current_best_feature is not None:
        best_features.append(current_best_feature)
        mse_history.append(best_score)

print("\nSelected Top 10 Features:", best_features)
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pymatgen.core import Structure
from pymatgen.analysis.local_env import VoronoiNN
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_squared_error

class StructureFeaturizer:

    def __init__(self):
        self.vnn = VoronoiNN(allow_pathological=True)

    def _electronegativity(self, st, stats=['min', 'max', 'mean']):
        data = [element.X for element in st.composition.elements if element.X is not None]
        if not data:
            data = [0.0]
        features = {}
        for stat_name in stats:
            np_func = getattr(np, stat_name)
            features[f"elneg_{stat_name}"] = float(np_func(data))
        return features

    def _atomic_properties(self, st, stats=['min', 'max', 'mean']):
        weights = [element.atomic_mass for element in st.composition.elements]
        numbers = [element.Z for element in st.composition.elements]
        features = {}
        for stat_name in stats:
            np_func = getattr(np, stat_name)
            features[f"weight_{stat_name}"] = float(np_func(weights))
            features[f"Z_{stat_name}"] = float(np_func(numbers))
        return features

    def _geometrical_voronoi(self, st, stats=['min', 'max', 'mean']):
        cn_list = []
        for i in range(len(st)):
            try:
                cn = self.vnn.get_cn(st, i, use_weights=True)
                cn_list.append(cn)
            except Exception:
                cn_list.append(0.0)
        if not cn_list:
            cn_list = [0.0]
        features = {}
        for stat_name in stats:
            np_func = getattr(np, stat_name)
            features[f"voronoi_cn_{stat_name}"] = float(np_func(cn_list))
        # Также добавим объем на один атом
        features["volume_per_atom"] = float(st.volume / len(st))
        return features

    def featurize(self, st, stats=['min', 'max', 'mean']):
        features = {}
        features.update(self._electronegativity(st, stats=stats))
        features.update(self._atomic_properties(st, stats=stats))
        features.update(self._geometrical_voronoi(st, stats=stats))
        return features

    def featurize_many(self, structures, stats=['min', 'max', 'mean']):
        features_list = []
        for st in structures:
            features_list.append(self.featurize(st, stats=stats))
        return pd.DataFrame(features_list)
````
